# Ejercicio 1 - Administración de Propietarios

#Ejercicio 2 - Pokédex de Movimientos

```python
# =========================================
# POKÉDEX DE MOVIMIENTOS POKÉMON
# =========================================

pokemon = {
    "Pikachu": ["Impactrueno", "Ataque Rapido"],
    "Charmander": ["Ascuas", "Arañazo"],
    "Squirtle": ["Pistola Agua", "Placaje"]
}


# =========================================
# FUNCIONES
# =========================================

def mostrar_pokedex(pokemon):

    for nombre, movimientos in pokemon.items():

        print("\n", nombre)

        for movimiento in movimientos:
            print("-", movimiento)


def agregar_movimiento(pokemon):

    nombre = input("Pokémon: ")
    movimiento = input("Movimiento: ")

    pokemon[nombre].append(movimiento)

    print("Movimiento agregado.")


def mostrar_movimientos(pokemon):

    nombre = input("Pokémon: ")

    print("\nMovimientos:")

    for movimiento in pokemon[nombre]:
        print("-", movimiento)


def contar_movimientos(pokemon):

    nombre = input("Pokémon: ")

    print(nombre, "posee", len(pokemon[nombre]), "movimientos.")


def agregar_pokemon(pokemon):

    nombre = input("Nombre del Pokémon: ")
    movimiento = input("Primer movimiento: ")

    pokemon[nombre] = [movimiento]

    print("Pokémon agregado.")


def eliminar_movimiento(pokemon):

    nombre = input("Pokémon: ")
    movimiento = input("Movimiento a eliminar: ")

    pokemon[nombre].remove(movimiento)

    print("Movimiento eliminado.")


# =========================================
# PRUEBAS
# =========================================

mostrar_pokedex(pokemon)

agregar_movimiento(pokemon)

mostrar_movimientos(pokemon)

contar_movimientos(pokemon)

agregar_pokemon(pokemon)

eliminar_movimiento(pokemon)

print("\nPOKÉDEX ACTUALIZADA")
mostrar_pokedex(pokemon)
```

# Ejercicio 3 - Registro de videojuegos

```python
# ============================
# PIXEL STORE
# GESTION DE VIDEOJUEGOS
# ============================

videojuegos = {}


# ============================
# FUNCIONES
# ============================

def registrar_videojuego(videojuegos):

    codigo = input("Ingrese código: ")

    if codigo in videojuegos:
        print("Código ya registrado.")
    else:
        nombre = input("Ingrese nombre: ")
        genero = input("Ingrese género: ")

        videojuegos[codigo] = {
            "nombre": nombre,
            "genero": genero
        }

        print("Videojuego agregado correctamente.")


def buscar_videojuego(videojuegos):

    codigo = input("Ingrese código a buscar: ")

    if codigo in videojuegos:
        print("\nCódigo:", codigo)
        print("Nombre:", videojuegos[codigo]["nombre"])
        print("Género:", videojuegos[codigo]["genero"])
    else:
        print("Videojuego no encontrado.")


def eliminar_videojuego(videojuegos):

    codigo = input("Ingrese código a eliminar: ")

    if codigo in videojuegos:
        del videojuegos[codigo]
        print("Videojuego eliminado.")
    else:
        print("No se pudo eliminar.")


def mostrar_catalogo(videojuegos):

    if len(videojuegos) == 0:
        print("No hay videojuegos registrados.")
    else:

        print("\n===== CATÁLOGO =====")

        for codigo in videojuegos:

            print("\nCódigo:", codigo)
            print("Nombre:", videojuegos[codigo]["nombre"])
            print("Género:", videojuegos[codigo]["genero"])


# ============================
# PROGRAMA PRINCIPAL
# ============================

while True:

    print("\n===== PIXEL STORE =====")
    print("1. Registrar videojuego")
    print("2. Buscar videojuego")
    print("3. Eliminar videojuego")
    print("4. Mostrar catálogo")
    print("5. Salir")

    try:

        opcion = int(input("Seleccione una opción: "))

        if opcion == 1:
            registrar_videojuego(videojuegos)

        elif opcion == 2:
            buscar_videojuego(videojuegos)

        elif opcion == 3:
            eliminar_videojuego(videojuegos)

        elif opcion == 4:
            mostrar_catalogo(videojuegos)

        elif opcion == 5:
            print("Programa finalizado.")
            break

        else:
            print("Debe ingresar una opción válida.")

    except:
        print("Error. Debe ingresar un número.")
```

# Ejercicio 4 - Team Builder de LoL

```python
# =========================================
# TEAM BUILDER - LEAGUE OF LEGENDS
# =========================================

campeones = {
    "Garen":{"roles":["TOP"],"winrate":51},
    "Darius":{"roles":["TOP"],"winrate":52},
    "Sett":{"roles":["TOP","JG"],"winrate":50},
    "Malphite":{"roles":["TOP"],"winrate":53},
    "Aatrox":{"roles":["TOP"],"winrate":49},

    "Lee Sin":{"roles":["JG"],"winrate":49},
    "Vi":{"roles":["JG"],"winrate":52},
    "Jarvan IV":{"roles":["JG"],"winrate":51},
    "Viego":{"roles":["JG"],"winrate":50},
    "Wukong":{"roles":["TOP","JG"],"winrate":52},

    "Ahri":{"roles":["MID"],"winrate":52},
    "Yasuo":{"roles":["MID","TOP"],"winrate":49},
    "Zed":{"roles":["MID"],"winrate":50},
    "Veigar":{"roles":["MID"],"winrate":53},
    "Lux":{"roles":["MID","BOTTOM"],"winrate":51},

    "Jinx":{"roles":["BOTTOM"],"winrate":52},
    "Vayne":{"roles":["TOP","BOTTOM"],"winrate":56},
    "Ashe":{"roles":["BOTTOM"],"winrate":51},
    "Caitlyn":{"roles":["BOTTOM"],"winrate":53},
    "Miss Fortune":{"roles":["BOTTOM"],"winrate":52},

    "Thresh":{"roles":["BOTTOM"],"winrate":51},
    "Nautilus":{"roles":["BOTTOM"],"winrate":52},
    "Leona":{"roles":["BOTTOM"],"winrate":50},
    "Morgana":{"roles":["MID","BOTTOM"],"winrate":51},
    "Pyke":{"roles":["MID","BOTTOM"],"winrate":53}
}

equipo_azul = {}
equipo_rojo = {}


# =========================================
# FUNCIONES
# =========================================

def campeon_ya_usado(nombre):
    return nombre in equipo_azul.values() or nombre in equipo_rojo.values()


def seleccionar_campeon(rol):

    while True:

        nombre = input(f"Ingrese campeón para {rol}: ")

        if nombre not in campeones:
            print("Campeón no encontrado.")
            continue

        if campeon_ya_usado(nombre):
            print("Ese campeón ya fue seleccionado.")
            continue

        if rol not in campeones[nombre]["roles"]:
            print(f"{nombre} no puede jugar {rol}.")
            continue

        return nombre


def crear_equipo(nombre_equipo):

    equipo = {}

    print(f"\n===== CREANDO {nombre_equipo} =====")

    equipo["TOP"] = seleccionar_campeon("TOP")
    equipo["JG"] = seleccionar_campeon("JG")
    equipo["MID"] = seleccionar_campeon("MID")
    equipo["BOTTOM1"] = seleccionar_campeon("BOTTOM")
    equipo["BOTTOM2"] = seleccionar_campeon("BOTTOM")

    return equipo


def mostrar_equipo(nombre, equipo):

    print(f"\n===== {nombre} =====")

    if len(equipo) == 0:
        print("Equipo no creado.")
        return

    for rol, campeon in equipo.items():
        print(f"{rol}: {campeon}")


def calcular_promedio(equipo):

    suma = 0

    for campeon in equipo.values():
        suma += campeones[campeon]["winrate"]

    promedio = suma / 5

    return promedio


def calcular_ganador():

    if len(equipo_azul) != 5 or len(equipo_rojo) != 5:
        print("Debe crear ambos equipos primero.")
        return

    promedio_azul = calcular_promedio(equipo_azul)
    promedio_rojo = calcular_promedio(equipo_rojo)

    print(f"\nPromedio Equipo Azul: {promedio_azul:.1f}%")
    print(f"Promedio Equipo Rojo: {promedio_rojo:.1f}%")

    if promedio_azul > promedio_rojo:
        print("El Equipo Azul tiene mayores probabilidades de ganar.")

    elif promedio_rojo > promedio_azul:
        print("El Equipo Rojo tiene mayores probabilidades de ganar.")

    else:
        print("Ambos equipos tienen las mismas probabilidades de victoria.")


# =========================================
# MENU PRINCIPAL
# =========================================

while True:

    try:

        print("\n===== TEAM BUILDER =====")
        print("1.- Crear Equipo Azul")
        print("2.- Crear Equipo Rojo")
        print("3.- Mostrar Equipos")
        print("4.- Calcular Ganador Probable")
        print("5.- Salir")

        opcion = int(input("Ingrese opción: "))

        if opcion == 1:

            equipo_azul = crear_equipo("EQUIPO AZUL")

        elif opcion == 2:

            equipo_rojo = crear_equipo("EQUIPO ROJO")

        elif opcion == 3:

            mostrar_equipo("EQUIPO AZUL", equipo_azul)
            mostrar_equipo("EQUIPO ROJO", equipo_rojo)

        elif opcion == 4:

            calcular_ganador()

        elif opcion == 5:

            print("Programa terminado...")
            break

        else:

            print("Debe ingresar una opción válida.")

    except:

        print("Error. Debe ingresar un número.")
```

# Ejercicio 5 - Registro planetario

```python
# =========================================
# SUPER TIERRA
# REGISTRO DE MUNDOS COLONIZADOS
# =========================================

planetas = {}


# =========================================
# FUNCIONES
# =========================================

def obtener_tipo_planeta(poblacion):

    if poblacion == 0:

        print("Tipo asignado automáticamente: Muerto")
        return "Muerto"

    elif poblacion <= 100000000:

        while True:

            print("\nLa población permite los siguientes tipos:")
            print("1. Mortífero")
            print("2. Salvaje")

            try:

                opcion = int(input("Seleccione tipo: "))

                if opcion == 1:
                    return "Mortífero"

                elif opcion == 2:
                    return "Salvaje"

                else:
                    print("Debe seleccionar una opción válida.")

            except:
                print("Debe ingresar un número.")

    else:

        while True:

            print("\nLa población permite los siguientes tipos:")
            print("1. Colmena")
            print("2. Forja")
            print("3. Agrícola")

            try:

                opcion = int(input("Seleccione tipo: "))

                if opcion == 1:
                    return "Colmena"

                elif opcion == 2:
                    return "Forja"

                elif opcion == 3:
                    return "Agrícola"

                else:
                    print("Debe seleccionar una opción válida.")

            except:
                print("Debe ingresar un número.")


def registrar_planeta(planetas):

    codigo = input("Ingrese código planetario: ")

    if codigo in planetas:
        print("El código ya existe.")
        return

    nombre = input("Ingrese nombre del planeta: ")

    while True:

        try:

            poblacion = int(input("Ingrese población: "))

            if poblacion >= 0:
                break

            print("La población no puede ser negativa.")

        except:
            print("Debe ingresar un número válido.")

    tipo = obtener_tipo_planeta(poblacion)

    planetas[codigo] = {
        "nombre": nombre,
        "tipo": tipo,
        "poblacion": poblacion
    }

    print("Planeta registrado con éxito.")


def buscar_planeta(planetas):

    codigo = input("Ingrese código del planeta: ")

    if codigo in planetas:

        print("\n===== INFORMACIÓN DEL PLANETA =====")
        print("Código:", codigo)
        print("Nombre:", planetas[codigo]["nombre"])
        print("Tipo:", planetas[codigo]["tipo"])
        print("Población:", planetas[codigo]["poblacion"])

    else:
        print("Planeta no encontrado.")


def actualizar_poblacion(planetas):

    codigo = input("Ingrese código del planeta: ")

    if codigo not in planetas:
        print("Planeta no encontrado.")
        return

    while True:

        try:

            nueva_poblacion = int(input("Ingrese nueva población: "))

            if nueva_poblacion >= 0:
                break

            print("La población no puede ser negativa.")

        except:
            print("Debe ingresar un número válido.")

    planetas[codigo]["poblacion"] = nueva_poblacion
    planetas[codigo]["tipo"] = obtener_tipo_planeta(nueva_poblacion)

    print("Población actualizada correctamente.")


def eliminar_planeta(planetas):

    codigo = input("Ingrese código del planeta: ")

    if codigo in planetas:

        del planetas[codigo]

        print("Planeta eliminado.")

    else:
        print("No se pudo eliminar el planeta.")


def mostrar_colonias(planetas):

    if len(planetas) == 0:
        print("No existen planetas registrados.")
        return

    print("\n===== COLONIAS REGISTRADAS =====")

    for codigo in planetas:

        print("\nCódigo:", codigo)
        print("Nombre:", planetas[codigo]["nombre"])
        print("Tipo:", planetas[codigo]["tipo"])
        print("Población:", planetas[codigo]["poblacion"])


# =========================================
# MENÚ PRINCIPAL
# =========================================

while True:

    print("\n===== SUPER TIERRA =====")
    print("1. Registrar planeta")
    print("2. Buscar planeta")
    print("3. Actualizar población")
    print("4. Eliminar planeta")
    print("5. Mostrar colonias")
    print("6. Salir")

    try:

        opcion = int(input("Seleccione una opción: "))

        if opcion == 1:
            registrar_planeta(planetas)

        elif opcion == 2:
            buscar_planeta(planetas)

        elif opcion == 3:
            actualizar_poblacion(planetas)

        elif opcion == 4:
            eliminar_planeta(planetas)

        elif opcion == 5:
            mostrar_colonias(planetas)

        elif opcion == 6:
            print("Programa finalizado.")
            break

        else:
            print("Debe ingresar una opción válida.")

    except:
        print("Error. Debe ingresar un número.")
```
